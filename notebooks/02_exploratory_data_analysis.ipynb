```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set visual styling for clean portfolio charts
sns.set_theme(style="whitegrid")
plt.rcParams['figure.figsize'] = (10, 5)

# Load cleaned master sales data
df = pd.read_csv('../data/processed/cleaned_master_sales.csv')

# Ensure timestamp columns are parsed as datetime
date_cols = ['order_purchase_timestamp', 'order_delivered_customer_date', 'order_estimated_delivery_date']
for col in date_cols:
    df[col] = pd.to_datetime(df[col])

print(f"Dataset Shape: {df.shape}")
df.head()
```

## 💡 Executive Summary & Key Business Insights

### 1. Revenue Trends & Seasonality
* **Observation:** Revenue peaked significantly in November (Black Friday spike) and maintained steady growth throughout 2017–2018.
* **Business Impact:** High Q4 order volumes create fulfillment and inventory pressures.
* **Recommendation:** Prepare inventory and supplier agreements 60 days before Q4 to prevent stockouts.

### 2. Top Product Categories
* **Observation:** Categories like `health_beauty`, `bed_bath_table`, and `watches_gifts` drive over 25% of overall sales.
* **Business Impact:** Heavy reliance on a few primary product lines.
* **Recommendation:** Expand seller partnerships in top categories while cross-selling related accessories.

### 3. Delivery & Shipping Performance
* **Observation:** Average delivery takes ~12 days, but remote states (like northern Brazil) take over 20+ days.
* **Business Impact:** Long delivery times directly lead to lower customer review scores.
* **Recommendation:** Partner with local regional fulfillment hubs in distant states to speed up delivery times.
